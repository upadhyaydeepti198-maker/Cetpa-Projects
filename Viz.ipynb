```python
import matplotlib.pyplot as plt
fruit_sale = [100,200,300,400]
month = ["Jan","Feb","Mar","APR"]
plt.figure(figsize = (5,8))
plt.bar(month,fruit_sale,color = ["red","gray","yellow","black"],hatch = "+")
plt.title("title")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,12,8,14,7]
plt.plot(x,y)
```

```python
study_hours = [1,2,3,4,5]
exam_scores = [55,60,70,75,85]
plt.scatter(study_hours,exam_scores,color="black",alpha=0.2,)
plt.title("Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()
```

```python
labels = ["Apple","Samsung","MicroSoft","Google"]
sale = [25,35,30,10]
plt.figure(figsize = (6,6))
plt.pie(sale,labels = labels,autopct = "%1.1f%%",shadow=True)
plt.title("Companies and Sales")
plt.show()
```

```python
fig,ax = plt.subplots(2,2,figsize=(10,8))
ax[0,0].bar(["Jan","Feb","Mar"],[100,200,300])
ax[0,0].set_title("Bar")

ax[0,1].bar(["Jan","Feb","Mar"],[100,200,300])
ax[0,1].set_title("Bar")

labels = ["Apple","Samsung","MicroSoft","Google"]
sale = [25,35,30,10]
ax[1,0].pie(sale,labels = labels,autopct = "%1.1f%%",shadow=True)
ax[1,0].set_title("Pie")
```

```python
months = ["Jan","Feb","March","APL","May","June",
          "July","Aug","Sept","Oct","Nov","Dec"]
sales_24= [50,55,53,60,70,75,72,68,70,74,80,85]
sales_25 = []
for i in sales_24:
  sales_25.append(i+10)
plt.plot(months,sales_24,label = "Sales 2024",color="blue",marker = "o")
plt.plot(months,sales_25,label = "Sales 2025",color="green",marker = "s")
plt.title("Sales Monthly Trend 2024 vs 2025")
plt.xlabel("Month")
plt.ylabel("Sale")
plt.xticks(rotation=45, ha='right')
plt.show()
```
